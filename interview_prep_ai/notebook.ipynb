# Voice Test

### pip install SpeechRecognition pyaudio pyttsx3

Make sure the notebook kernel is using the venv interpreter at `venv/Scripts/python.exe`.

If the interpreter is wrong, VS Code may not find `pywintypes` even though it is installed in the venv.

```python
import speech_recognition as sr
r = sr.Recognizer()

while True:
    try:
        with sr.Microphone() as source:
            print("Listening...")
            
            r.adjust_for_ambient_noise(source, duration=0.2)
            audio = r.listen(source)
            text = r.recognize_google(audio)
            text = text.lower()  
            print("You said:", text)
            
            if "exit" in text:
                print("Exiting program...")
                break

    except sr.RequestError as e:
        print("Could not request results; {0}".format(e))

    except sr.UnknownValueError:
        print("Could not understand audio")

    except KeyboardInterrupt:
        print("Program terminated by user")
        break
```

```python
import sys
print('Python executable:', sys.executable)

try:
    import pyttsx3
    engine = pyttsx3.init('sapi5')
except Exception as e:
    print('TTS initialization failed:', type(e).__name__, e)
    raise

text = input('Enter the text you want to convert to speech: ')

engine.say(text)
engine.runAndWait()

engine.save_to_file(text, 'output_audio.mp3')
engine.runAndWait()

print("Text has been spoken and saved as 'output_audio.mp3'")
```
