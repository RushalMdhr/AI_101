```python
import ollama
import json
```

```python
def ai(msg):
    response = ollama.chat(model='llama3.2:latest', messages=msg, 
                           options={
                                "temperature": 0.9,      # Creative but not wild
                                # "top_p": 0.9,            # Standard
                                # "num_predict": 200,      # Short story (~200 tokens)
                                # "num_ctx": 4096,         # Enough for context
                                # "repeat_penalty": 1.1,   # Avoid repetition
                                # "seed": 123,             # Reproducible
                                # "mirostat": 1            # Adaptive creativity
                            },
                            format='json'
                            )
    return response['message']['content']
```

```python
def user_talking(text):
    return {"role": "user", "content": text}
def system_talking(text):
    return {"role": "system", "content": text}
```

```python
message=[]
prompt="""
You are a keyword extractor. From the user's text, extract the most important keywords.

Rules:
- Return ONLY valid JSON. No explanation, no markdown.
- Format: {"keywords": ["word1", "word2", ...]}
- Extract 3-6 keywords: topics, skills, tools, roles, or key entities.
- Use lowercase, single words or short phrases (2-3 words max).
- Ignore filler words (I, me, my, the, am, an).

Example:
user: my name is roshan. i dont like football. but i love ronaldo alot
ai: {"keywords": ["roshan", "football", "ronaldo"]}
"""
message.append(system_talking(prompt))
message.append(user_talking("I'm an AI assistant built to help you learn faster — I explain concepts in short, structured answers and adapt to your preferred modes (normal vs explore). I'm good at coding help, AI/ML concepts, debugging, and clear explanations without fluff. Think of me as a fast, no-nonsense tutor who keeps answers tight unless you ask for depth."))
```

```python
res = ai(message)
```
