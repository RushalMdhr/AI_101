```python
def update_prompt():
    print("updating your prompt")
    choice = input("want to type prompt yourself? Y/N : ").lower()
    if choice =="y":
        system_prompt = input("write your prompt : ")
    else:
        system_prompt = """
        You Are an agent with given tools to calculate numbers
        so each function is defined and to use the function u need to respond in this way
        {"tool_name": [arg1, arg2]}
        Tool names available are [useMultiplier,useAdder]
        if user tells u to multiple useMutilier and if user says u to add useAdder
        """

    return system_prompt
```

```python
import ollama

def useOllama(messages):
    return ollama.chat(model='llama3.2:latest', messages=messages)
```

```python
import json

messages = [{"role": "system", "content": "You are a helpful assistant"}]

while True:
    user_input = input("You: ")
    if user_input.lower() == "exit":
        break
    elif user_input.lower() == "prompt":
        system_prompt = update_prompt()
        messages[0]["content"] = system_prompt
        continue

    # Add user message
    messages.append({"role": "user", "content": user_input})
    
    # Get bot response
    response = useOllama(messages)
    bot_says = response['message']['content']
    
    # Handle JSON tool call
    if bot_says.startswith("{"):
        try:
            fixed_json = bot_says.replace("'", '"')  # Fix single quotes
            data = json.loads(fixed_json)
            key = list(data.keys())[0]
            value = data[key]
            
            if key == "useMultiplier":
                result = multiplication(value[0], value[1])
                # Append tool result as a separate message
                messages.append({"role": "assistant", "content": bot_says})
                messages.append({"role": "tool", "name": "multiplier", "content": str(result)})
                # Get final response with tool result
                final_response = useOllama(messages)
                bot_says = final_response['message']['content']
        except json.JSONDecodeError:
            print("⚠️ Invalid JSON from bot, using raw response")
    
    # Append final bot message
    messages.append({"role": "assistant", "content": bot_says})
    print(f"Bot: {bot_says}")
```
