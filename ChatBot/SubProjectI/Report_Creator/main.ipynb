```python
import os
import gradio as gr
from groq import Groq
from dotenv import load_dotenv
load_dotenv()
# print(os.getenv("GROQ_API_KEY"))
client = Groq(api_key=os.getenv("GROQ_API_KEY"))  # Replace or use env var
```

```python
def generate_report(topic):
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[{"role": "user", "content": f"Write a detailed report on: {topic}"}]
    )
    return response
    # return response.choices[0].message.content

# demo = gr.Interface(
#     fn=generate_report,
#     inputs=gr.Textbox(lines=2, label="Enter Report Topic"),
#     outputs=gr.Textbox(lines=15, label="Your Report"),
#     title="AI Report Generator"
# )
# try:
#     demo.launch()
# except Exception as er:
#     print(er)
```
