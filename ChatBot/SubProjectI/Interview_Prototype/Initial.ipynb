### 1. Category distribution

```python
df['category'].value_counts().plot(kind='bar')
plt.title('Questions by Category')
plt.show()
```

```python
# Create pivot table
pivot = pd.crosstab(df['experience'], df['difficulty'])
# Plot heatmap with seaborn
plt.figure(figsize=(8,6))
sns.heatmap(pivot, annot=True, cmap='YlOrRd', fmt='d')
plt.title('Experience vs Difficulty Heatmap - HR QUESTIONS')
plt.show()
```
